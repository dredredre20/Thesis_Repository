```python
import os
import torch

# This will contain the edge values (placeholder values for now)
edge_index = torch.tensor([
    [0, 1, 0, 2, 0, 4, 2, 4], # integers represent the user
    [1, 0, 2, 0, 4, 0, 4, 2], # ex. user 1 is connected to user 0, is connected to user 2, and so on
], dtype=torch.long)

# This will contain the node values (each row represents a user)
node_features = torch.tensor([
    [25, 1], # format: [age, 1 = likes sports; 0 = does not like sports]
    [30, 0],
    [22, 1],
    [35, 0],
    [27, 1],
], dtype=torch.float)
```

**Visualizing GNNs**

```python
import networkx as nx
import matplotlib.pyplot as plt

edge_list = edge_index.t().tolist() # convert to list

# Create a NetworkX graph
G = nx.Graph()
G.add_edges_from(edge_list)

G.add_nodes_from(range(node_features.size(0))) # for isolated nodes

pos = nx.spring_layout(G, seed=42) # fixed seed

# Draw a graph
plt.figure(figsize=(6,6))
nx.draw_networkx(G, pos, with_labels=True, node_color='lightblue', edge_color='gray', node_size=800)
plt.title("Visualization of the Social Network Graph")
plt.axis('off')
plt.show()
```
