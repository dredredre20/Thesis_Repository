## Start Getting Posts

```python
import requests
import time

API_URL = "https://public.api.bsky.app/xrpc/app.bsky.feed.getPosts"
BATCH_SIZE = 25

def fetch_hydrated_posts(uris, batch_size=BATCH_SIZE, pause=1):
    """
    Fetch hydrated posts from the Bluesky API in batches.
    """

    hydrated_posts = []
    for i in range(0, len(uris), batch_size):
        batch_uris = uris[i:i + batch_size]
        response = requests.get(API_URL, params={"uris": batch_uris})
        if response.status_code == 200:
            hydrated_posts.extend(response.json().get('posts', []))
            time.sleep(pause)
        else:
            print(f"Error fetching posts for URIs {batch_uris}: {response.status_code}")
        time.sleep(1)  # To avoid hitting rate limits
    return hydrated_posts
```

```python

URIs = bsky_sample_pd['uri'].tolist()

hydrated_posts = fetch_hydrated_posts(URIs)
print(f"Fetched {len(hydrated_posts)} hydrated posts.")
```

```python
# Extract all quoted posts from the hydrated posts

quoted_posts = [
    p["embed"]["record"]
    for p in hydrated_posts
    if p.get("embed", {}).get("$type") == "app.bsky.embed.record#view"
    and p["embed"]["record"].get("$type") == "app.bsky.embed.record#viewRecord"
]

print(f"Found {len(quoted_posts)} quoted posts.")
```

Get all users who reposted the post

```python
REPOSTS_URL = "https://public.api.bsky.app/xrpc/app.bsky.feed.getRepostedBy"

def get_reposters(uri, limit=100, pause=1.0):
    """
    Fetch the list of users who reposted a given post URI.
    """
    reposters, cursor = [], None

    while True:
        params = {"uri": uri, "limit": limit}

        if cursor:
            params["cursor"] = cursor
        resp = requests.get(REPOSTS_URL, params=params)
        resp.raise_for_status()
        data = resp.json()
        reposters.extend(data["repostedBy"])
        cursor = data.get("cursor")

        if not cursor:
            break
        time.sleep(pause)

    return reposters
```

Get a view of the author's feed composed of posts and reposts

```python
AUTHOR_FEED_URL = "https://public.api.bsky.app/xrpc/app.bsky.feed.getAuthorFeed"

def get_author_feed(actor, max_posts=10, filter="posts_with_replies", pause=1.0):
    """Return up to max_posts feed items for an actor (DID or handle)."""
    items, cursor = [], None
    while len(items) < max_posts:

        params = {"actor": actor, "limit": 100, "filter": filter}

        if cursor:
            params["cursor"] = cursor
        resp = requests.get(AUTHOR_FEED_URL, params=params)

        if resp.status_code != 200:
            print(f"Error for {actor}: {resp.status_code} {resp.text}")
            break

        data = resp.json()
        items.extend(data["feed"])
        cursor = data.get("cursor")

        if not cursor:
            break
        time.sleep(pause)
    return items[:max_posts]
```

Get the accounts that an acccount follows

```python
FOLLOW_URL = "https://public.api.bsky.app/xrpc/app.bsky.graph.getFollows"

def get_follows(actor, limit=10, pause=1.0):
    """Return up to limit follows for an actor (DID or handle)."""
    follows, cursor = [], None
    while len(follows) < limit:

        params = {"actor": actor, "limit": limit}

        if cursor:
            params["cursor"] = cursor
        resp = requests.get(FOLLOW_URL, params=params)

        if resp.status_code != 200:
            print(f"Error for {actor}: {resp.status_code} {resp.text}")
            break

        data = resp.json()
        follows.extend(data["follows"])
        cursor = data.get("cursor")

        if not cursor:
            break
        time.sleep(pause)
    return follows[:limit]
```

```python
sample_follows = {}
sample_reposters = reposters[:10]

for user in sample_reposters:
    sample_follows[user["did"]] = get_follows(user["did"], limit=10)
```

```python
def make_date_readable(date_str):
    """Convert ISO 8601 date string to a more readable format."""
    from datetime import datetime
    dt = datetime.fromisoformat(date_str.replace("Z", "+00:00"))
    return dt.strftime("%Y-%m-%d %H:%M:%S")
```
